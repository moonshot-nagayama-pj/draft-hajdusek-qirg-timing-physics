```python
import numpy as np
import matplotlib.pyplot as plt
```

## Polarization

```python
theta = np.linspace(-np.pi/2, np.pi/2, 201)
coincidence = np.sin(theta)**2/2
visibility = np.cos(theta)**2
```

```python
fig, ax = plt.subplots(figsize=(5,3.5))

ax.plot(theta, visibility, lw=2, color='black', label=r'visibility')
ax.plot(theta, coincidence, lw=2, ls='--', color='black', label=r'coincidence prob.')

ax.set_ylim(-0.1,1.3)
ax.set_xlabel(r'Polarization mismatch', fontsize=12)
ax.set_xticks([-np.pi/2, 0, np.pi/2])
ax.set_xticklabels([r'$-\pi/2$', 0, r'$\pi/2$'], fontsize=10)
ax.set_yticks([0, 0.5, 1.0])
ax.set_yticklabels(['0.0', '0.5', '1.0'], fontsize=10)
ax.legend(ncols=2, frameon=False, fontsize=10)

plt.savefig('visibility-polarization.svg', transparent=True, bbox_inches='tight')
plt.show()
```

## Spectral distinguishability

```python
frequency_difference = np.linspace(-6,6,101)
std_ratio = np.linspace(0, 10, 501)

coincidence_freq = 0.5 * (1 - np.exp(-frequency_difference**2/2))
visibility_freq = 1 - 2 * coincidence_freq

coincidence_std = 0.5 - std_ratio / (1 + std_ratio**2)
visibility_std = 2 * std_ratio / (1 + std_ratio**2)
```

```python
fig, ax = plt.subplots(2,1, figsize=(5,7))

ax[0].plot(frequency_difference, visibility_freq, lw=2, color='black', label=r'visibility')
ax[0].plot(frequency_difference, coincidence_freq, lw=2, color='black', ls='--', label=r'coincidence prob.')
ax[1].plot(std_ratio, visibility_std, lw=2, color='black', label=r'visibility')
ax[1].plot(std_ratio, coincidence_std, lw=2, color='black', ls='--', label=r'coincidence prob.')

ax[0].set_ylim(-0.1,1.3)
ax[0].set_xlabel(r'Central frequency mismatch', fontsize=12)
ax[0].set_title('Case A', fontsize=12)
ax[0].set_yticks([0, 0.5, 1.0])
ax[0].set_yticklabels(['0.0', '0.5', '1.0'], fontsize=10)
ax[0].legend(ncols=2, frameon=False, fontsize=10)

ax[1].set_ylim(-0.1,1.3)
ax[1].set_xlabel(r'Ratio of standard deviations', fontsize=12)
ax[1].set_title('Case B', fontsize=12)
ax[1].set_yticks([0, 0.5, 1.0])
ax[1].set_yticklabels(['0.0', '0.5', '1.0'], fontsize=10)
ax[1].legend(ncols=2, frameon=False, fontsize=10)

plt.tight_layout()
plt.savefig('visibility-spectral.svg', transparent=True, bbox_inches='tight')
plt.show()
```

## Temporal distinguishability

```python
time_diff = np.linspace(-6,6,101)

coincidence = 0.5 * (1 - np.exp(-0.5 * time_diff**2))
visibility = np.exp(-0.5 * time_diff**2)
```

```python
fig, ax = plt.subplots(figsize=(5,3.5))

ax.plot(time_diff, visibility, lw=2, color='black', label=r'visibility')
ax.plot(time_diff, coincidence, lw=2, ls='--',color='black', label=r'coincidence prob.')

ax.set_ylim(-0.1,1.3)
ax.set_xlabel(r'Time delay', fontsize=12)
ax.set_yticks([0, 0.5, 1.0])
ax.set_yticklabels(['0.0', '0.5', '1.0'], fontsize=10)
ax.legend(ncols=2, frameon=False, fontsize=10)

plt.savefig('visibility-temporal.svg', transparent=True, bbox_inches='tight')
plt.show()
```
