```python
import numpy as np
import matplotlib.pyplot as plt
```

# Spectral distinguishability

```python
frequency = np.linspace(-15,15,201)

meanA = -2.5
meanB = 2.5
varA = 7
varB = 7
meanC = 0
meanD = 0
varC = 15
varD = 1

photonA = 1/(np.sqrt(2*np.pi*varA)) * np.exp(-((frequency - meanA)**2)/varA)
photonB = 1/(np.sqrt(2*np.pi*varB)) * np.exp(-((frequency - meanB)**2)/varB)
overlapAB = np.minimum(photonA, photonB)

photonC = 1/(np.sqrt(2*np.pi*varC)) * np.exp(-((frequency - meanC)**2)/varC)
photonD = 1/(np.sqrt(2*np.pi*varD)) * np.exp(-((frequency - meanD)**2)/varD)
overlapCD = np.minimum(photonC, photonD)
```

```python
fig, ax = plt.subplots(2, 1, figsize=(7,7))

fig.subplots_adjust(hspace=0.5) 

ax[0].plot(frequency, photonA, color='black')
ax[0].plot(frequency, photonB, color='black')
ax[0].fill_between(frequency, overlapAB, fc="w", hatch="//")

ax[0].spines['left'].set_visible(False)
ax[0].spines['right'].set_visible(False)
ax[0].spines['top'].set_visible(False)

ax[0].set_xticks([])
ax[0].set_yticks([])
ax[0].set_xlabel('Frequency', fontsize=12)
ax[0].set_title('Different central frequencies', fontsize=12)

ax[1].plot(frequency, photonC, color='black')
ax[1].plot(frequency, photonD, color='black')
ax[1].fill_between(frequency, overlapCD, fc="w", hatch="//")

ax[1].spines['left'].set_visible(False)
ax[1].spines['right'].set_visible(False)
ax[1].spines['top'].set_visible(False)

ax[1].set_xticks([])
ax[1].set_yticks([])
ax[1].set_xlabel('Frequency', fontsize=12)
ax[1].set_title('Different standard deviations', fontsize=12)

plt.savefig('distinguishability-spectral.svg', transparent=True, bbox_inches='tight')
plt.show()
```

# Temporal distinguishability

```python
frequency = np.linspace(-15,15,201)

meanA = -1.5
meanB = 1.5
varA = 4
varB = 4

meanC = -7.5
meanD = 7.5
varC = 4
varD = 4

photonA = 1/(np.sqrt(2*np.pi*varA)) * np.exp(-((frequency - meanA)**2)/varA)
photonB = 1/(np.sqrt(2*np.pi*varB)) * np.exp(-((frequency - meanB)**2)/varB)
overlapAB = np.minimum(photonA, photonB)

photonC = 1/(np.sqrt(2*np.pi*varC)) * np.exp(-((frequency - meanC)**2)/varC)
photonD = 1/(np.sqrt(2*np.pi*varD)) * np.exp(-((frequency - meanD)**2)/varD)
```

```python
fig, ax = plt.subplots(2, 1, figsize=(7,7))

fig.subplots_adjust(hspace=0.5) 

ax[0].plot(frequency, photonA, color='black')
ax[0].plot(frequency, photonB, color='black')
ax[0].fill_between(frequency, overlapAB, fc="w", hatch="//")

ax[0].spines['left'].set_visible(False)
ax[0].spines['right'].set_visible(False)
ax[0].spines['top'].set_visible(False)

ax[0].set_xticks([])
ax[0].set_yticks([])
ax[0].set_xlabel('Time', fontsize=12)
ax[0].set_title('Partially distinguishable', fontsize=12)

ax[1].plot(frequency, photonC, color='black')
ax[1].plot(frequency, photonD, color='black')

ax[1].spines['left'].set_visible(False)
ax[1].spines['right'].set_visible(False)
ax[1].spines['top'].set_visible(False)

ax[1].set_xticks([])
ax[1].set_yticks([])
ax[1].set_xlabel('Time', fontsize=12)
ax[1].set_title('Fully distinguishable', fontsize=12)

plt.savefig('distinguishability-temporal.svg', transparent=True, bbox_inches='tight')
plt.show()
```
